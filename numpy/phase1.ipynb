# comparison of list vs array

```python
#list multiplication
py_list=[1,2,3]
print("list multiplication:",py_list*2)

#array multiplication
np_arr=np.array([1,2,3])#here element wise multiplication occur
print("list multiplication:",np_arr*2)


#note-array me har ek element ke sath multiplication hota hai lekin list me har ek element double show honne lagta hai 
```

# creating array from scratch 

```python
#first we create an array whose all elements are zero
zeros=np.zeros((2,3))
print ("matrix of zeros:\n",zeros)

#now we create an array whose all elements are ones
ones=np.ones((2,3))
print ("matrix of ones:\n",ones)

#now we create an array whose all elements are any of the constant value with the help of full method 
full=np.full((2,3),8)
print ("matrix of constant value :\n",full)

#now we create an array whose all elements are any of the random value with the help of random class and random  method 
random=np.random.random((2,3))
print ("matrix of ones:\n",random)
```

# creating vector ,matrix and tensor 

```python
#vector
vector=np.array([1,2,3,4,5,6,])
print("vector:",vector)

#matrix
matrix=np.array([[1,2,3,4],[3,4,5,6]])
print("\nmatrix:\n",matrix)

#tensor
tensor=np.array([[[1,2],[3,4]],
                 [[4,5],[6,7]]])
print("\n tensor:/n",tensor)
```

## array properties 

```python
arr=np.array([[1,2,3],[4,5,6]])
print("Shape:",arr.shape)  #gives rows and column
print("dimension:",arr.ndim)#gives dimension
print("size",arr.size)#size method gives us total number of elements in the array 
print("datatype",arr.dtype)#this method gives the datatype of the array
```

## reshaping the array 

```python
arr=np.arange(12)
print("original array:",arr)

reshaped=arr.reshape(3,4)     #basically it reshaped the arr as per the given rows and columns
print("after rehshaping :",reshaped)

#if you want to do it again jaise pahle tha vaise then there is a method called flatten()
flattened=reshaped.flatten()
print("flattened array is",flattened)

#for transposed array
transposed=reshaped.T
print("transposed array is \n",transposed)
```
