#comparison of list vs array

```python
#list multiplication
py_list=[1,2,3]
print("list multiplication:",py_list*2)

#array multiplication
np_arr=np.array([1,2,3])#here element wise multiplication occur
print("list multiplication:",np_arr*2)


#note-array me har ek element ke sath multiplication hota hai lekin list me har ek element double show honne lagta hai 
```

# creating array from scratch 

```python
#first we create an array whose all elements are zero
zeros=np.zeros((2,3))
print ("matrix of zeros:\n",zeros)

#now we create an array whose all elements are ones
ones=np.ones((2,3))
print ("matrix of ones:\n",ones)

#now we create an array whose all elements are any of the constant value with the help of full method 
full=np.full((2,3),8)
print ("matrix of constant value :\n",full)

#now we create an array whose all elements are any of the random value with the help of random class and random  method 
random=np.random.random((2,3))
print ("matrix of ones:\n",random)
```
