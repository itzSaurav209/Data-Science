# NUMPY array operations 

```python
import numpy as np
```

# learning the concept of slicing in array numpy 

```python
arr=np.array([1,2,3,4,5,6,6,7,8,9,10])
print(arr)
print("Basic slicing",arr[2:7])
print("with step",arr[1:7:2])
print("negative indexing",arr[-3])
```

# now we will learn how to do indexing and slicing in 2_D array

```python
arr_2d=np.array([[2,3,4],
                 [4,5,8],
                 [6,7,8]])
#if we want element of 2 row of 2 column 
print("specific element",arr_2d[1,1])
print("specific row only without column",arr_2d[2])
print("specific column only",arr_2d[:,2])
```

# sorting
### sorting in 1d and 2d array both 

```python
unsorted=np.array([0,2,3,62,1,2,4,5,36])
print("sorted array:",np.sort(unsorted))

#2d array
unsorted_2d=np.array([[1,2],[3,2],[6,5]])
#now there are two types of sorting in 2d array -: 1)row wise 2)column wise 

#first we will see column wise sorting
print("column wise sorting",np.sort(unsorted_2d,axis=0))    #axis =0 means column wise sorting  
print("row wise sorting",np.sort(unsorted_2d,axis=1))    #axis =0 means row wise sorting  
 
```
